## Worked examples

### Particle on a spring
A particle of mass $m$ is fixed to a spring (with spring constant $k$ and rest length $l_0$). The spring rests on a horizontal table, see figure. At $t=0$, the spring has its rest length. The particle is released with zero velocity.
What is the motion of the particle?

```{figure} ../images/WorkedEx6.1.svg
:width: 30%
:alt: a point mass fixed on top of a vertical positioned spring that rests on a table.

```

::::{tab-set}
:::{tab-item} Interpret the problem

First we finish the sketch by drawing what is relevant for this problem.

```{figure} ../images/WorkedEx6.1a.svg
:width: 30%
:alt: a point mass fixed on top of a vertical positioned spring that rests on a table.

```
The motion of the particle is governed by gravity and the force exerted by the spring. Both act along the $z$-axis, making this a 1-dimensional problem.
We have a particle that is initially at rest and will start moving when it is released.
::: 

:::{tab-item} Develop the solution

We will set up N2 and specify the initial conditions.

$$m\frac{dv}{dt} = -mg - k(l-l_0)$$

Note that the force from the spring has a negative sign! This needed to get the direction of the force right: if the length of the spring is stretched beyond its rest length, the particle is pulled down.

We have two initial conditions:

$$\begin{split}
v(0) = 0 \\
z(0) = l
\end{split}$$

::: 
:::{tab-item} Evaluate the problem

We can write N2 in terms of the position $z$ of the particle:

$$m\frac{d^2z}{dt^2} + k(z-l_0) = -mg$$

This is a second order differential equation. We can solve it by first finding the homogeneous solution and then a particular solution.

The homogeneous equation is:

$$m\frac{d^2z}{dt^2} + k(z-l_0) = 0$$

with solution:

$$z_h(t) = A\cos(\omega t) + B\sin(\omega t)$$

where $\omega = \sqrt{\frac{k}{m}}$.

Next we need a particular or special solution. We can find is by trying to find the steady state (or equilibrium) position. Thus we set $\frac{d^2z}{dt^2} = 0$.

$$k(z-l_0) = -mg \rightarrow z_{eq} = l_0 - \frac{mg}{k}$$

Thus the general solution is

$$z(t) = z_h(t) + z_{eq} = A\cos(\omega t) + B\sin(\omega t) + l_0 - \frac{mg}{k}$$

Now we apply the initial conditions:

$$\begin{split}
z(0) = A + l_0 - \frac{mg}{k} = l_0 \rightarrow A = \frac{mg}{k}\\
v(0) = -A\omega \sin(0) + B\omega \cos(0) = B\omega = 0 \rightarrow B = 0
\end{split}$$

And we find as solution

$$z(t) = \frac{mg}{k}\cos(\omega t) + l_0 - \frac{mg}{k}$$

with 

$$\omega = \sqrt{\frac{k}{m}}$$

::: 
:::{tab-item} Assess the problem

We see, that the particle will start oscillating as we expected. Moreover, the particle will first start moving downwards. This makes sense, as at $t=0$ gravity pulls down the mass, but the spring does not exert an upward force as it is at its rest length.

The particle start oscillating around $z_{eq} = l_0 - \frac{mg}{k}$. This makes sense: if we would have placed the particle at $z_{eq}$ without initial velocity, it would stay there as the two forces acting on it would then exactly cancel each other.

::: 
::::



```{exercise} Chaos
:label: ex_chaos

A famous example of oscillations is the following. Consider the equation:

$$x_{i+1} = r \cdot x_i \cdot (1-x_i)$$

with $0<r<4$ and $0<x_0<1$. 

For values of $r<3$, $x$ converges to a single value. For values for $r$ between ~3 and ~3.4. the value of x oscillates. If we further increase the value of $r$ their will be period-doubling, eventually leading to chaos.

Write the code that plots $x$ as function of $i$ and skip the first 10 steps of $i$ to better see convergence. Try to change the value for $r$ and see the outcome.

In for a challenge? Try to reproduce this [bifurcation diagram](https://en.wikipedia.org/wiki/File:Logistic_Map_Bifurcation_Diagram,_Matplotlib.svg)

```







## Answers
```{solution} ex_61
:class: dropdown

a. equilibrium: $\sum F = 0$
$$0 = -mg + V_p \rho_{oil} g - kz_{eq} \rightarrow z_{eq} = \frac{V_p (\rho_p - \rho_{oil} ) g}{k}$$

with $V_p$ the particle volume and $\rho_p$ the density of the particle.

b. N2: now we have to take the frictional force by the oil into account:

$$ m\frac{d^2z}{dt^2} = -kz - V_p (\rho_p - \rho_{oil} ) g - b\frac{dz}{dt} $$

c. Rewrite this as 

$$m\ddot{z} + b \dot{z} + kz = - V_p (\rho_p - \rho_{oil} ) g$$

This equation describes a damped harmonic oscillator. It will oscillate sinusoidal with angular frequency $\omega_\pm = \frac{-3\pi\mu R \pm \sqrt{9\pi^2\mu^2 R^2 - mk}}{m}$

The general solution is:

$$z(t) = z_{eq} + A \sin \omega t + B \cos \omega t$$

```

````{solution} ex_62
:class: dropdown

a. 
```{figure} ../images/Exercise6.2.svg
:width: 30%
:alt: Plot of the potential energy.

```

b. To find the equilibrium positions, we need to find the points where the derivative of the potential energy is zero:
$$\frac{dV}{dx} = \frac{c(a^2 - x^2)}{(x^2 + a^2)^2} = 0 \rightarrow x = \pm a$$

The stability follows from inspecting 
$$\frac{d^2V}{dx^2} = -2cx \frac{-x^2 + 3a^2}{x^2 + a^2}^3$$

$$ x=-a \rightarrow \frac{d^2V}{dx^2} = \frac{c}{2a^3} > 0 \rightarrow \text{stable}$$
$$ x=a \rightarrow \frac{d^2V}{dx^2} = -\frac{c}{2a^3} < 0 \rightarrow \text{unstable}$$

The same conclusion can also be drawn by inspecting the graph of $V(x)$.

c. Make a Taylor expansion around $x=-a$:

$$ V(x) = V(-a) + \frac{1}{1!} V'(-a) (x+a) +\frac{1}{2!} V"(-a) (x+a)^2 + h.o.t.$$

with $V'(-a)=0$ and $V"(-a)=\frac{c}{2a^3}$.
Thus, we can write the energy for a small interval around $x=-a$ as:

$$\frac{1}{2}mv^2 + \frac{1}{2}\frac{c}{2a^3} (x+a)^2 -\frac{c}{2a} = E_0$$

This is the energy of a harmonic oscillator with angular frequency $\omega = \sqrt{\frac{c}{2a^3 m}}$ and period $T = 2\pi \sqrt{\frac{2a^3 m}{c}}$.

````

````{solution} ex_63
:class: dropdown

a. 
```{figure} ../images/Exercise6.3a.svg
:width: 30%
:alt: Plot of forces acting on m.

```
2 forces, both Coulomb type:<br> 
left charge $F_L$, right charge $F_R$<br>
position: $\Delta x = x$<br>
left charge: $F_L = \frac{1}{4\pi\epsilon_0}\frac{Qq}{(L+ x )^2}$<br>
right charge: $F_R = -\frac{1}{4\pi\epsilon_0}\frac{Qq}{(L- x )^2}$  
    
equilibrium: 
$$\begin{split}
\sum F_i = 0 \rightarrow F_L + F_R = 0 \rightarrow \\
\frac{1}{4\pi\epsilon_0}\frac{Qq}{(L+ x )^2} - \frac{1}{4\pi\epsilon_0}\frac{Qq}{(L- x )^2} = 0 \rightarrow x = 0 \rightarrow\\
(L-x)^2 = (L+x)^2 \rightarrow x_{eq} = 0
\end{split}$$  	


b. Energy conservation!!
Coulomb force → potential exists: same relation with distance as gravity
$$F_L = \frac{1}{4\pi\epsilon_0}\frac{Qq}{(L+ x )^2} \rightarrow V_L(x) = -\int_\infty^x \frac{1}{4\pi\epsilon_0}\frac{Qq}{(L+ x )^2} dx = \frac{1}{4\pi\epsilon_0}\frac{Qq}{L+ x}$$
$$F_R = -\frac{1}{4\pi\epsilon_0}\frac{Qq}{(L- x )^2} \rightarrow V_R(x) = -\int_\infty^x -\frac{1}{4\pi\epsilon_0}\frac{Qq}{(L- x )^2} dx = \frac{1}{4\pi\epsilon_0}\frac{Qq}{L- x}$$ 

So, energy conservation:
$$\begin{split}
\frac{1}{2}mv^2 +V_L(x) + V_R(x) = E_0 \rightarrow \\
\frac{1}{2}mv^2 + \frac{1}{4\pi\epsilon_0}\frac{Qq}{L+ x} + \frac{1}{4\pi\epsilon_0}\frac{Qq}{L- x} =E_0
\end{split}$$

```{figure} ../images/Exercise6.3b.svg
:width: 60%
:alt: Plot of potential energy.

``` 
From a plot of the potential energy we see: the motion of $m$ is confined! Moreover, $m$ will oscillate, but probably not in a harmonic way!

c. Equation of motion:

$$ m\frac{d^2x}{dt^2} = \frac{1}{4\pi\epsilon_0}\frac{Qq}{(L+ x )^2} - \frac{1}{4\pi\epsilon_0}\frac{Qq}{(L- x )^2}$$

d. Rewrite N2 using $s=x/L$:
$$mL\frac{d^2s}{dt^2} = \frac{Qq}{4\pi\epsilon_0 L^2} \left (\frac{1}{(1+s)^2} - \frac{1}{(1-s)^2} \right )$$

e. Make a Taylor series for the forces around $s=0$:

$$\begin{split}
\frac{1}{1+s^2} \approx 1 -2s + 3s^2 \\
\frac{1}{1-s^2} \approx 1 +2s + 3s^2
\end{split}$$

Thus, for $s \ll 1$$ we find in good approximation:

$$\begin{split}
mL\frac{d^2s}{dt^2} = \frac{Qq}{4\pi\epsilon_0 L^2} \cdot -4s \\
\frac{d^2s}{dt^2} + \frac{Qq}{\pi\epsilon_0 mL^3} s = 0
\end{split}$$

harmonic oscillator!!

f. with frequency
$$\omega_0^2 = \frac{Qq}{\pi\epsilon_0 mL^3}$$
````


````{solution} ex_64
:class: dropdown

a. Kirchhoff: $V_L + V_R + V_C = 0 \rightarrow L\frac{dI}{dt} + IR + \frac{1}{C}\int Idt = 0$ 

b. relation charge and current: $ q = \int Idt \leftrightarrow I = \frac{dq}{dt} \rightarrow \frac{dI}{dt} = \frac{d^2q}{dt^2}$

Thus, we can write for the charge:

$$ L\frac{d^2q}{dt} + R\frac{dq}{dt} +\frac{1}{C}q = 0$$

Damped harmonic oscillator!<br>
‘mass’: $L$, thus the coil is the mass=inertia<br>
spring constant: $\frac{1}{C}$, hence the capacitor is the spring, that generates the restoring force<br>
friction: $R$, the resistor is the damping.<br>
$q(t)$ acts as ‘position coordinate’

c. with driving voltage source:

$$ V_L + V_R +V_C = V_0 \sin \nu t \rightarrow L\frac{d^2q}{dt} + R\frac{dq}{dt} +\frac{1}{C}q = V_0 \sin \nu t$$

This describes a forced, damped harmonic oscillator with general solution:

$$q(t) = A e^{-\frac{R}{2L}t} \sin \left (\sqrt{\frac{1}{LC} - \frac{R^2}{4L^2}}t + \epsilon \right ) + q_{max} \sin (\nu t + \alpha )$$

d. with 
$$q_{max} = \frac{V_0 / L}{\sqrt{(\omega_0^2 - \nu^2)^2 + \frac{R^2}{L^2} \nu^2}}$$

which has a maximum if the denominator is minimal, i.e. $\nu_{res}^2 = \omega_0^2 - \frac{R^2}{L^2}$ and $\omega_0^2 = \frac{1}{LC}$

Thus, for $R=10$Ohm, $L=0.1$H and $C=1.0 \cdot 10^{-5}$F: $\nu_{res}$ = 994 rad/s


````
